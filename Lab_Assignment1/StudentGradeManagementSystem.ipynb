```python
# Classes and Function (3 marks)

# Step 1: Create a class BankAccount.
# Add class attribute bankname="Canada Bank"
# Add 2 instance attributes: account_holder (string) & balance (float, default value 0.0)
class BankAccount:
    bankname = "Canada Bank"
    def __init__(self, account_holder, balance=0.0):
        self.account_holder = account_holder
        self.balance = balance

    # Step 2: Add methods
    # 1. deposit(amount) -> Add amount to the balance
    # 2. withdraw(amount) -> Subtract the amount if balance is sufficient, else print an error message
    # 3. get_balance() -> Print Account holder's name and current balance
    # 4. get_bankname() -> Print the class attribute (staticmethod)
    # Note: Use f-strings to display the amount and updated balance whenever you call deposit() or withdraw()
    def deposit(self, amount):
        self.balance += amount
        print(f"[{BankAccount.bankname}] Deposited ${amount}. New balance: ${self.balance}")

    def withdraw(self, amount):
        if self.balance >= amount:
            self.balance -= amount
            print(f"[{BankAccount.bankname}] Withdrew ${amount}. Remaining balance: ${self.balance}")
        else:
            print(f"[{BankAccount.bankname}] Insufficient funds. Current balance: ${self.balance}")

    def get_balance(self):
        print(f"[{BankAccount.bankname}] Account Holder: {self.account_holder}, Current Balance: ${self.balance}")

    @staticmethod
    def get_bankname():
        print(f"Bank Name: {BankAccount.bankname}")


# Step 3: Test
# 1. Print the Bankname
# 2. Create an object of BankAccount with your name and an initial balance of 1000
# 3. Deposit 500
# 4. Withdraw 200
# 5. Print the final balance
BankAccount.get_bankname()
account = BankAccount("Sam Steel", 1000)
account.deposit(500)
account.withdraw(200)
account.get_balance()
```

```python
# step 1
# import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# step 2
# Load data.xlsx file, which contains the observations (with noise), variance of observations and true observations
# hint: use pandas library to parse excel file and use numpy library to convert it to array
# (1 mark)
df_data = pd.read_excel('data.xlsx')
arr_data = df_data.to_numpy()

# Print the columns to identify the correct column names
print(df_data.columns)

#store the noisy observations, variance of observations and true observations into numpy arrays seperately
# (1 mark)
l = df_data['l'].to_numpy()
l_var = df_data['l_var'].to_numpy()
l_true = df_data['l_true'].to_numpy()
```

```python
# step 5
# define a curve fitting function named fit_polynomial_curve
# the function should implement the linear least square estimation
# the function should include all important intermediate steps in LSE
# the input and output of the funciton are described in the comment below in red.
# please keep code clean and well-commented
# (7 marks)

def fit_polynomial_curve(a: np.ndarray, l: np.ndarray, P: np.ndarray, deg: int) -> tuple:
    '''
    Args:
        a:              function input (constants)
        l:              noisy observations
        P:              observation weight matrix
        deg:            degree of polynomial to fit to data
    Returns:
        x:              estimated polynomial coefficients
        cov_x:          uncertainty in estimated coefficients
        apv:            a posterier variance factor (goodness of fit)
        v:              observation residuals
    '''

    # Number of observations
    n = l.shape[0]

    # Design matrix A
    # The design matrix depends on the degree of the polynomial.
    # For a polynomial of degree 'deg', the columns will be a^0, a^1, ..., a^deg
    A = np.vander(a, deg + 1, increasing=True)

    # Normal matrix N = A^T * P * A
    N = A.T @ P @ A

    # Right-hand side vector b = A^T * P * l
    b = A.T @ P @ l

    # Estimated coefficients x = N^-1 * b
    x = np.linalg.solve(N, b)

    # Estimated observations l_hat = A * x
    l_hat = A @ x

    # Observation residuals v = l_hat - l
    v = l_hat - l

    # A posteriori variance factor (apv)
    # apv = (v^T * P * v) / (n - number of unknowns)
    # Number of unknowns is the number of coefficients, which is deg + 1
    apv = (v.T @ P @ v) / (n - (deg + 1))

    # Covariance matrix of estimated coefficients cov_x = apv * N^-1
    cov_x = apv * np.linalg.inv(N)


    return (x, cov_x, apv, v)
```

Key Conclusions: When the reliability (precision) of the observed data varies, weighted least squares regression outperforms unweighted least squares regression in terms of curve fitting quality and parameter estimation accuracy.

Key Observations:

Lower Posterior Variance Factor:

For all tested polynomial orders (2, 4, and 6), the posterior variance factor of weighted regression is significantly lower than that of unweighted regression.

This indicates that weighted regression provides a better overall fit to the data.

Smaller Norm Between Predicted and True Values:

The norm between the predicted and true observations in weighted regression is smaller, especially in high-order polynomial models.

This indicates that the predicted curve obtained by weighted regression is closer to the true observed data points and has higher fitting accuracy.

Explanation and Principle:

The fundamental reason for the discrepancy lies in whether the uncertainty information of the observations is utilized.

Unweighted regression uses a unit weight matrix and treats all data points equally. It cannot distinguish between high-precision and low-precision (noisy) observations, resulting in results that are easily biased by unreliable data points.

Weighted regression: A weight matrix P (typically the inverse of the observation covariance matrix) is used to treat data points differently.

Observations with small variance (high precision) are given greater weight, giving them a greater impact on the final fitting result.

Observations with large variance (low precision) are given smaller weights to reduce the impact of their noise on the results.

Advantages of weighted regression:

Noise suppression: The negative impact of unreliable data points is automatically reduced.

Improves estimation accuracy: The resulting model parameters (polynomial coefficients) are more likely to be close to their true values.

Provides more reliable uncertainty estimates: The parameter covariance matrix more accurately reflects the uncertainty in the data.

In summary, weighted least squares (WLS) achieves a smarter and more robust fit by incorporating data uncertainty information. It is a statistically superior method for handling heteroskedastic data (where observations have unequal variances).
