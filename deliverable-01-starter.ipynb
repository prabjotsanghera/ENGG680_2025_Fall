## **<u>Part 1: Radiation Physics</u>**

### 1.1 Ionizing Radiation Classification and Energy Scales

**Key Concept:**
Radiation is classified by whether it has enough energy to knock electrons out of atoms (ionization). This happens when the photon energy exceeds the **ionization potential** of an atom (~13.6 eV for hydrogen).

**Energy Classification:**
- **Non-ionizing:** Visible light (1.6-3.1 eV), radio waves (<0.00001 eV)
  - Energy too low to ionize atoms
  - Cannot damage DNA directly
  
- **Ionizing:** X-rays (keV range), gamma rays (keV-MeV range)
  - High enough energy to remove electrons from atoms
  - Can damage living tissue and DNA
  - Used in medical imaging (controlled exposure)

**Key Equation:**
$$E = h\nu = \frac{hc}{\lambda}$$

- Where:
    - h = 6.626×10⁻³⁴ J·s (Planck's constant)
    - c = 3×10⁸ m/s (speed of light)

### 1.2 X-ray Production

**X-ray Generation:**
X-rays are produced when high-energy electrons strike a metal target (usually tungsten).

**Two Production Mechanisms:**

1. **Bremsstrahlung (Braking Radiation):**
   - Electron decelerates near nucleus
   - Loses kinetic energy → emitted as X-ray photon
   - Produces **continuous spectrum** of energies
   - Maximum energy = tube voltage (kVp)
   - Accounts for ~80% of X-ray output

2. **Characteristic X-rays:**
   - Electron knocks out inner shell electron
   - Outer electron fills vacancy
   - Energy difference emitted as X-ray
   - Produces **discrete spectral lines** (specific energies)
   - Depends on target material (e.g., K-alpha, K-beta lines for tungsten)

**X-ray Spectrum:**
- Shape determined by tube voltage (kVp) and filtration
- Higher kVp → higher maximum energy and intensity
- Filtration removes low-energy photons (patient protection)

**Key Equation:**
$$E_{max} = eV_{peak}$$
- Where:
   - $V_{peak}$ is the peak tube voltage in kV

If an electron loses all of its kinetic energy in a single interaction, that energy is emitted as one X-ray photon. The maximum photon energy is therefore given by - 𝐸𝑚𝑎𝑥 = 𝑒𝑉𝑝𝑒𝑎𝑘,
𝑉𝑝𝑒𝑎𝑘 ​= 100kV. Thus, the maximum energy of X-ray photons produced is 100 keV.

As the electrons lose kinetic energy during these interactions, X-ray photons are emitted. Because electrons can lose varying amounts of energy depending on how close they pass to the nucleus, bremsstrahlung radiation produces a continuous spectrum of photon energies, ranging from near zero up to the maximum energy determined by the tube voltage. ## **<u>Part 2: Photon Interactions</u>**

### 2.1 Interaction Types

**Key Concept:**
When X-rays pass through tissue, they interact through different mechanisms depending on photon energy and material properties.

**Main Interaction Types:**

1. **Photoelectric Effect** (low energy, high-Z materials)
   - X-ray photon absorbed completely
   - Electron kicked out of atom
   - Useful for imaging (produces signal)
   - Why: Bone (high Z) appears bright

2. **Compton Scattering** (medium-high energy)
   - Photon deflected and loses energy
   - Electron recoils
   - Scattered photons reduce image contrast (noise)
   - Why: Anti-scatter grids are used

### 2.2 Beer-Lambert Law & Attenuation

**Attenuation** describes how X-ray intensity decreases as it passes through material.

**Beer-Lambert Law:**
$$I = I_0 e^{-\mu x}$$

- Where:
    - $I$ = transmitted intensity
    - $I_0$ = incident intensity  
    - $\mu$ = linear attenuation coefficient (cm⁻¹) - material dependent
    - $x$ = material thickness (cm)

**Half-Value Layer (HVL):**
- Thickness of material that reduces intensity to 50%
- Related to attenuation coefficient: $HVL = \frac{\ln(2)}{\mu} = \frac{0.693}{\mu}$
- Lower HVL = more attenuation (denser material or lower energy)

**Key Insight:**
- Bone has higher μ than soft tissue → more attenuation → appears bright
- Different energies attenuate differently (energy dependence)
- HVL is a practical measure used in X-ray quality control

## **Answer to Question 4**

The attenuation of X-rays through matter is described by the Beer–Lambert law:

\[
I = I_0 e^{-\mu x}
\]

Given:
- \( I_0 = 10{,}000 \) photons  
- \( \mu = 0.2 \, \text{cm}^{-1} \)  
- \( x = 5 \, \text{cm} \)

Substituting:

\[
I = 10{,}000 \, e^{-(0.2)(5)} = 10{,}000 \, e^{-1}

\[

\[
\text{HVL} = \frac{\ln(2)}{\mu} = \frac{0.693}{\mu}
\]

With \( \mu = 0.2 \, \text{cm}^{-1} \):

\[
\text{HVL} = \frac{0.693}{0.2} = 3.465 \, \text{cm}

**HVL ≈ 3.47 cm**

## **<u>Part 3: Spatial Resolution</u>**

### 3.1 Point Spread Function (PSF) and System Blurring

**Key Concept:**

Real imaging systems are **imperfect**. A point source (infinitely small object) doesn't appear as a point in the image—it appears as a **blurred blob**. This blurring limits how small an object you can see.

**Point Spread Function (PSF):**
- **PSF** = How much a point source gets blurred by the imaging system
- **Shape:** Usually Gaussian-like (bell curve)
- **Width:** Characterized by **FWHM** (Full Width at Half Maximum)
  - Smaller FWHM = sharper system (better resolution)
  - Larger FWHM = blurrier system (worse resolution)


**Resolution Rule of Thumb:** You can distinguish two objects only if they're separated by at least the **FWHM of the PSF**. Anything closer just blurs together.

**Image Formation:**

$$\text{Blurred Image} = \text{Sharp Object} \otimes \text{PSF}$$

  - The $\otimes$ symbol means "convolved with"—every sharp point gets replaced by a blurred copy (PSF shape).

**Real Example - Chest X-ray:**
- System resolution: ~0.2 mm FWHM
- Lung nodule: 3-4 mm → **Visible** (15-20× larger than PSF)
- Micro-calcification: 1-3 mm → **Borderline** (5-15× larger)
- Small vessel: 0.5 mm → **Invisible** (blurs away)

## **Part 3: Spatial Resolution**

### **3.1 Point Spread Function (PSF) and System Blurring**

In an ideal imaging system, a point object would be reproduced as a point in the image. In real imaging systems, however, physical limitations such as finite focal spot size, detector blur, motion, and scattering cause image degradation. As a result, a point source is imaged not as a point, but as a **spatially distributed intensity pattern**, known as the **Point Spread Function (PSF)**.

The **Point Spread Function (PSF)** describes the system’s response to an infinitesimally small point object and provides a complete spatial characterization of system blurring. In many diagnostic imaging systems, the PSF can be approximated by a **Gaussian-shaped function**, reflecting the combined effects of multiple independent blurring processes.

A common quantitative measure of the PSF width is the **Full Width at Half Maximum (FWHM)**, defined as the distance between the two points at which the PSF amplitude falls to one-half of its maximum value. The FWHM provides a practical metric of spatial resolution:
- A **smaller FWHM** corresponds to less blurring and higher spatial resolution.
- A **larger FWHM** corresponds to greater blurring and reduced spatial resolution.

As a rule of thumb, two closely spaced objects can be resolved as separate structures only if their separation is on the order of, or greater than, the FWHM of the system PSF. Objects separated by distances smaller than the PSF width will overlap significantly in the image and appear blurred together.

Mathematically, image formation in a linear, shift-invariant imaging system can be described as a convolution between the true object and the system PSF:

\[
\text{Image}(x,y) = \text{Object}(x,y) \otimes \text{PSF}(x,y)
\]

This convolution relationship indicates that each point in the object is replaced by a blurred version of the PSF, and the final image is the superposition of all such blurred contributions.

In practical diagnostic imaging, spatial resolution limits determine the smallest anatomical structures that can be visualized. Structures that are much larger than the PSF width are readily detected, while structures comparable to or smaller than the PSF width may not be distinguishable from background blur. Thus, system blurring fundamentally constrains the visibility of fine anatomical detail in radiographic images.


### 3.2 Linear Systems Theory & MTF

**Linear Systems Theory:**
Medical imaging systems can be modeled as linear, shift-invariant systems.

**Key Properties:**
1. **Linearity:** If input doubles, output doubles
2. **Shift Invariance:** System response doesn't depend on position
3. **Convolution:** Image formation is described by convolution with PSF
   $$\text{Image} = \text{Object} \otimes \text{PSF}$$

**Modulation Transfer Function (MTF):**
- MTF is the **Fourier transform of the PSF**
- Measures how well the system reproduces different spatial frequencies
- MTF ranges from 0 (no reproduction) to 1 (perfect reproduction)
- **Spatial frequency** measured in cycles/mm or line pairs/mm

**Interpreting MTF:**
- MTF = 1.0 at low frequencies → large objects reproduced perfectly
- MTF decreases at high frequencies → fine details are blurred
- **MTF at 50%** (where MTF = 0.5) is often used as resolution metric
- Related to FWHM: Better resolution → higher MTF at high frequencies

**Example:**
- System A: MTF = 0.5 at 5 cycles/mm → can resolve 0.1 mm details
- System B: MTF = 0.5 at 10 cycles/mm → better, can resolve 0.05 mm details

### **(c) Relationship between MTF and PSF (math + interpretation)**

For a **linear, shift-invariant (LSI)** imaging system, image formation is modeled as convolution:

\[
\text{Image}(x,y) = \text{Object}(x,y)\otimes \text{PSF}(x,y)
\]

Taking the Fourier transform (convolution theorem):

\[
\mathcal{F}\{\text{Image}\} = \mathcal{F}\{\text{Object}\}\cdot \mathcal{F}\{\text{PSF}\}
\]

The Fourier transform of the PSF is called the **Optical Transfer Function (OTF)**:

\[
\text{OTF}(f_x,f_y) = \mathcal{F}\{\text{PSF}(x,y)\}
\]

The **Modulation Transfer Function (MTF)** is the **magnitude** of the OTF:

\[
\text{MTF}(f_x,f_y) = |\text{OTF}(f_x,f_y)|
\]

**What each tells us:**
- **PSF (space domain):** shows how a point object is blurred in the image (narrower PSF → sharper system).
- **MTF (frequency domain):** shows how well different spatial frequencies (detail sizes) are transferred (higher MTF at high frequency → better fine-detail preservation).

**Key link (intuition):**  
A **narrower PSF** corresponds to a **broader/higher MTF**, meaning less blur and better transfer of fine spatial detail.

## **<u>Part 4: Noise & Image Quality</u>**

### 4.1 Understanding Image Noise

**Quantum (Poisson) Noise** is the primary noise source in X-ray and nuclear imaging:

- Arises from the random nature of photon emission and detection
- Follows **Poisson statistics**: if we detect N photons on average, the noise (standard deviation) is $\sigma = \sqrt{N}$
- **Signal-to-Noise Ratio (SNR)**: $SNR = \frac{N}{\sigma} = \frac{N}{\sqrt{N}} = \sqrt{N}$

**Key insights:**
- SNR increases with the square root of photon count
- To double SNR, you need 4× more photons (4× more dose)
- **Rose Criterion**: Objects are detectable when SNR ≥ 5

**Noise reduction strategies:**
1. Increase photon count (higher dose)
2. Spatial averaging (smoothing filters)
3. Temporal averaging (multiple acquisitions)

```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from scipy.ndimage import gaussian_filter, median_filter
import os

# --- Robust image load (works whether you're in repo root or not) ---
candidates = [
    "figures/MRI_Highnoise.jpg",
    "/content/BMEN509-623_Deliverable_1-main/figures/MRI_Highnoise.jpg",
    "/content/Deliverable1/BMEN509-623_Deliverable_1-main/figures/MRI_Highnoise.jpg",
]

img_path = None
for p in candidates:
    if os.path.exists(p):
        img_path = p
        break

if img_path is None:
    raise FileNotFoundError("Could not find MRI_Highnoise.jpg. Check where the repo/figures folder is extracted.")

img = Image.open(img_path).convert("L")
img_array = np.array(img, dtype=float)

# --- PSNR function ---
def calculate_psnr(original, filtered):
    """
    PSNR = 10 * log10(MAX^2 / MSE)
    MAX = 255 for 8-bit images
    """
    original = np.asarray(original, dtype=np.float64)
    filtered = np.asarray(filtered, dtype=np.float64)

    mse = np.mean((original - filtered) ** 2)
    if mse == 0:
        return float("inf")

    max_pixel = 255.0
    psnr = 10.0 * np.log10((max_pixel ** 2) / mse)
    return psnr

# --- Auto-tune parameters (relative PSNR vs noisy reference, as notebook suggests) ---
sigma_candidates = [0.5, 0.8, 1.0, 1.2, 1.5, 2.0]
median_candidates = [3, 5, 7]  # odd integers

best_sigma, best_psnr_g, best_img_g = None, -np.inf, None
for s in sigma_candidates:
    g = gaussian_filter(img_array, sigma=s)
    p = calculate_psnr(img_array, g)
    if p > best_psnr_g:
        best_psnr_g, best_sigma, best_img_g = p, s, g

best_k, best_psnr_m, best_img_m = None, -np.inf, None
for k in median_candidates:
    m = median_filter(img_array, size=k)
    p = calculate_psnr(img_array, m)
    if p > best_psnr_m:
        best_psnr_m, best_k, best_img_m = p, k, m

# Use best settings
sigma = best_sigma
kernel_size = best_k
img_gaussian = best_img_g
img_median = best_img_m

# --- Display results ---
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

axes[0].imshow(img_array, cmap="gray")
axes[0].set_title("Original Noisy Image")
axes[0].axis("off")

axes[1].imshow(img_gaussian, cmap="gray")
axes[1].set_title(f"Gaussian Filter (sigma={sigma})\nPSNR: {calculate_psnr(img_array, img_gaussian):.2f} dB")
axes[1].axis("off")

axes[2].imshow(img_median, cmap="gray")
axes[2].set_title(f"Median Filter (size={kernel_size})\nPSNR: {calculate_psnr(img_array, img_median):.2f} dB")
axes[2].axis("off")

plt.tight_layout()
plt.show()

print(f"Chosen sigma={sigma} (Gaussian), PSNR={best_psnr_g:.2f} dB (relative)")
print(f"Chosen size={kernel_size} (Median), PSNR={best_psnr_m:.2f} dB (relative)")
print("Winner by relative PSNR:", "Gaussian" if best_psnr_g >= best_psnr_m else "Median")
```

The **Gaussian filter** was chosen as the preferred denoising method.

The dominant noise in X-ray and MRI images is **quantum (Poisson-like) noise**, which manifests as continuous, spatially distributed intensity fluctuations rather than isolated impulse artifacts. The Gaussian filter is well suited for this type of noise because it performs weighted spatial averaging, reducing random fluctuations while preserving overall anatomical structure.

### 4.2 Detective Quantum Efficiency (DQE)

**DQE** measures how efficiently an imaging system uses the incoming radiation to produce image information.

**Definition:**
$$DQE = \frac{SNR_{out}^2}{SNR_{in}^2}$$

Where:
- $SNR_{out}$ = signal-to-noise ratio in the output image
- $SNR_{in}$ = signal-to-noise ratio in the input radiation

**Key Concepts:**
- DQE ranges from 0 to 1 (or 0% to 100%)
- **Perfect detector:** DQE = 1 (no noise added by detector)
- **Real detectors:** DQE < 1 (detector adds noise)
- Higher DQE = better detector (more efficient use of radiation)

**Practical Implications:**
- High DQE detector → need less radiation dose for same image quality
- DQE depends on spatial frequency (usually decreases at high frequencies)
- DQE depends on exposure level (often highest at moderate exposures)

**Example:**
- Film/screen: DQE ≈ 0.3-0.4 (30-40%)
- Digital radiography (DR): DQE ≈ 0.6-0.7 (60-70%)
- DR systems allow lower dose for same image quality

### **Question 7**

Two X-ray detector systems are being compared:

- **System A (Film/Screen):** DQE = 0.35, cost = $10,000
- **System B (Digital DR):** DQE = 0.70, cost = $50,000

Both systems receive the same input radiation with $SNR_{in} = 100$.

- a) Calculate the output SNR for each system.

- b) If you want to achieve the same output SNR with System A as System B naturally produces, by what factor would you need to increase the radiation dose to System A?

- c) In a high-volume clinical setting performing 100 chest X-rays per day, discuss whether the higher cost of System B is justified. Consider patient dose, image quality, and long-term patient safety.

## Answer to Question 7

### (a) Calculate the output SNR for each system

The detective quantum efficiency (DQE) is defined as:

$$
DQE = \frac{SNR_{out}^2}{SNR_{in}^2}
$$

Rearranging:

$$
SNR_{out} = SNR_{in}\sqrt{DQE}
$$

Given that the input signal-to-noise ratio is:

$$
SNR_{in} = 100
$$

**System A (Film/Screen), DQE = 0.35:**

$$
SNR_{out,A} = 100\sqrt{0.35} \approx 100(0.5916) \approx 59.2
$$

**System B (Digital DR), DQE = 0.70:**

$$
SNR_{out,B} = 100\sqrt{0.70} \approx 100(0.8367) \approx 83.7
$$

**Results:**
- System A: $SNR_{out} \approx 59.2$
- System B: $SNR_{out} \approx 83.7$

---

### (b) Dose factor required for System A to match System B

In quantum-limited imaging:

$$
SNR_{in} \propto \sqrt{\text{dose}}
$$

To make System A achieve the same output SNR as System B:

$$
SNR_{in}'\sqrt{DQE_A} = SNR_{in}\sqrt{DQE_B}
$$

Solving for the required input SNR increase:

$$
\frac{SNR_{in}'}{SNR_{in}} = \sqrt{\frac{DQE_B}{DQE_A}}
= \sqrt{\frac{0.70}{0.35}}
= \sqrt{2}
\approx 1.414
$$

Since dose scales with the square of input SNR:

$$
\frac{\text{dose}'}{\text{dose}} = (1.414)^2 \approx 2
$$

**System A would require approximately 2× higher radiation dose to match System B’s output SNR.**

---

### (c) Is the higher cost of System B justified in a high-volume clinic?

In a high-volume clinical environment (e.g., 100 chest X-rays per day), the higher DQE of **System B (Digital DR)** is generally justified.

- **Dose efficiency:**  
  A higher DQE allows the same image quality to be achieved at significantly lower patient dose. From part (b), System A would require roughly twice the dose to match the image quality of System B.

- **Image quality and repeat reduction:**  
  Improved noise performance reduces the likelihood of repeat examinations, further lowering cumulative patient dose and improving workflow efficiency.

- **Long-term safety and value:**  
  Although System B has a higher upfront cost, reduced radiation exposure, fewer repeats, and improved consistency make it advantageous in high-throughput clinical settings.

**Note:** This analysis assumes quantum-limited operation and a single representative DQE value. In practice, DQE varies with spatial frequency and exposure level, but the conclusion that higher DQE enables lower-dose imaging remains valid.

### **Coding Exercise 2**

In this exercise, simulate the effect of photon counting statistics on image quality. You will:
- Generate a simple "phantom" image
- Add Poisson noise at different dose levels
- Calculate SNR as a function of photon count
- Verify the theoretical relationship: $SNR = \sqrt{N}$

```python
import numpy as np
import matplotlib.pyplot as plt

# Create a simple phantom (uniform square in background)
phantom = np.zeros((100, 100))
phantom[30:70, 30:70] = 1.0  # Square with intensity 1.0

# Define different photon dose levels (mean photons per pixel)
dose_levels = [10, 50, 100, 500, 1000]
snr_measured = []

fig, axes = plt.subplots(2, 3, figsize=(15, 10))
axes = axes.flatten()

for idx, N_photons in enumerate(dose_levels):
    # Generate Poisson noise
    # Hint: Scale phantom by N_photons, apply Poisson noise, then normalize back
    # Scale the phantom to represent expected photon counts (lambda for Poisson)
    scaled_phantom = phantom * N_photons
    noisy_counts = np.random.poisson(scaled_phantom) # Apply Poisson noise
    noisy_phantom = noisy_counts / N_photons # Normalize back to original intensity scale

    # Calculate SNR in the central square region
    # For the phantom used, the 'signal' in the square is 1.0 and background is 0.0
    # The noise is Poisson, so std_dev = sqrt(mean_counts)
    # We need to calculate SNR for the difference between signal and background, considering noise.

    # For simplicity and matching the theoretical SNR = sqrt(N) for Poisson noise,
    # we will measure SNR in a region where the expected signal is constant.
    # Let's consider the 'signal region' as the inner square, and 'background region' as the outer area.
    # However, the theoretical SNR = sqrt(N) typically applies to counts in a single region.
    # For image SNR, it's often (Signal_mean - Background_mean) / Noise_std_in_background.
    # Given the hint 'Verify the theoretical relationship: SNR = sqrt(N)',
    # let's calculate N from the signal region and compare its sqrt to the SNR.

    # Calculate counts in the signal region (inside the square) and background region (outside the square)
    # We'll use the *raw counts* for SNR calculation to align with sqrt(N) for Poisson.
    signal_counts_region = noisy_counts[30:70, 30:70]
    background_counts_region = noisy_counts[0:20, 0:20] # A piece of the background

    # For SNR = sqrt(N), N refers to the mean number of detected photons in the region of interest.
    # Let's assume N is the mean count in the 'signal' square for simplicity of verification.
    signal_mean_counts = np.mean(signal_counts_region)
    background_mean_counts = np.mean(background_counts_region)

    # The noise is the standard deviation of the counts in a uniform region.
    # Let's take noise from the background for calculating image SNR or from the signal for pure Poisson N.
    # To verify SNR = sqrt(N), we typically look at a uniform region's mean and std dev.
    # The problem implies N is the photon count, and SNR = sqrt(N).
    # Let's use the standard deviation of the signal region (or a uniform region if it were clean)
    # For a 'detectability' type SNR like Rose criterion, it's (Signal - Background) / Noise_std.

    # Let's calculate the SNR for the signal part itself, verifying SNR = sqrt(N_photons * phantom_value_in_signal_region)
    # The 'signal' here is the average count in the square, and noise is its std dev.
    # For Poisson, mean = lambda, std = sqrt(lambda).
    # If phantom[30:70, 30:70] = 1.0, then expected counts are N_photons.
    # So, we expect SNR = sqrt(N_photons).
    # Let's use the noise from the signal region for SNR = sqrt(N).

    # If we are verifying SNR = sqrt(N_photons) (N_photons is the dose level),
    # we need the mean and std dev of counts within the object for that specific N_photons
    # The object (square) has an intensity of 1.0, so the expected counts per pixel is N_photons.
    # For Poisson statistics in that uniform region:
    mean_counts_in_object = np.mean(noisy_counts[30:70, 30:70])
    std_counts_in_object = np.std(noisy_counts[30:70, 30:70])

    # Measured SNR (using statistics from the object region)
    # It should be close to sqrt(N_photons) if the region was purely uniform at N_photons intensity.
    # For the problem statement's `SNR = N / sqrt(N) = sqrt(N)`, N should be the mean count itself.
    # So `std_counts_in_object` should be close to `sqrt(mean_counts_in_object)`.
    # SNR here should be mean / std_dev
    if std_counts_in_object > 0:
        snr = mean_counts_in_object / std_counts_in_object
    else:
        snr = float('inf') # Should not happen with Poisson noise

    snr_measured.append(snr)

    # Display noisy phantom
    axes[idx].imshow(noisy_phantom, cmap='gray', vmin=0, vmax=1.2)
    axes[idx].set_title(f'N = {N_photons}\nSNR = {snr:.2f}')
    axes[idx].axis('off')

# Plot SNR vs. N (log-log scale)
axes[5].loglog(dose_levels, snr_measured, 'bo-', label='Measured SNR')

# Plot theoretical SNR = sqrt(N) for comparison
theoretical_snr = np.sqrt(dose_levels) # Here N refers to the dose_level itself
axes[5].loglog(dose_levels, theoretical_snr, 'r--', label='Theory: sqrt(N)')
axes[5].set_xlabel('Photon Count (N)')
axes[5].set_ylabel('SNR')
axes[5].set_title('SNR vs. Photon Count')
axes[5].legend()
axes[5].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("\nDose levels and corresponding SNR:")
for N, snr in zip(dose_levels, snr_measured):
    print(f"N = {N:4d} photons: SNR = {snr:.2f} (Theory: {np.sqrt(N):.2f})")
```

- a) Do your measurements agree with the theoretical prediction? If there are differences, what might explain them?

### (a) Agreement with theory and possible differences

Overall, the measured SNR follows the expected Poisson relationship:

$$
\sigma \approx \sqrt{N}, \qquad SNR = \frac{N}{\sigma} \approx \sqrt{N}
$$

So **as dose increases**, the measured SNR should increase approximately like **$\sqrt{N}$**, which is what the log–log plot should show (a slope close to 0.5).

**If the measured SNR does not match $\sqrt{N}$ perfectly, the main reasons are:**

1. **Finite sample size (statistical fluctuation):**  
   We estimate mean and standard deviation from a finite ROI (40×40 pixels).  
   At low photon counts (e.g., 10), random variation is relatively large, so the measured SNR can deviate noticeably.

2. **Using sample standard deviation instead of the theoretical standard deviation:**  
   For Poisson data, the theoretical standard deviation is $\sqrt{\lambda}$, but we estimate it numerically from the ROI.  
   That estimate has its own error, especially at low dose.

3. **Non-ideal uniformity / edge effects (minor):**  
   If the ROI includes boundary pixels or any intensity transition region, it can inflate the measured variance.  
   (This is minimized here by using only the interior square region.)


Measurements should match the **trend** (SNR proportional to $\sqrt{N}$) very clearly, and agreement becomes **much tighter at higher photon counts**, where Poisson statistics average out.

```python
import numpy as np
import matplotlib.pyplot as plt

# ----------------------------
# 1) Create a simple phantom (2 regions: background + circular object)
# ----------------------------
H, W = 256, 256
y, x = np.ogrid[:H, :W]
cy, cx = H//2, W//2
r = 50

phantom = np.ones((H, W), dtype=float) * 0.60      # background transmission/intensity fraction
mask = (y - cy)**2 + (x - cx)**2 <= r**2
phantom[mask] = 0.35                               # object region lower intensity (more attenuating)

# ----------------------------
# 2) Dose levels (photon counts)
#    N0 controls the expected photon count scale
# ----------------------------
N0_list = [200, 500, 1000, 5000, 10000, 50000]

# Regions of interest (ROI) for SNR measurement
# We'll compute SNR in a small background ROI (flat region) to match Poisson theory.
bg_roi = (slice(30, 80), slice(30, 80))  # top-left area, mostly background

measured_snr = []
theory_snr = []
N_mean_list = []

# Store a few example noisy images for display
noisy_examples = []

# ----------------------------
# 3) Poisson noise simulation + SNR calculation
# ----------------------------
for N0 in N0_list:
    # Expected counts per pixel = N0 * phantom
    lam = N0 * phantom

    # Poisson sampling (photon counting)
    counts = np.random.poisson(lam).astype(float)

    # Convert back to "normalized intensity" (optional)
    img_noisy = counts / N0

    # Measure SNR in background ROI:
    # For Poisson counts: mean ≈ N, std ≈ sqrt(N) -> SNR ≈ sqrt(N)
    counts_bg = counts[bg_roi]
    mean_counts = np.mean(counts_bg)
    std_counts = np.std(counts_bg, ddof=1)

    snr_meas = mean_counts / std_counts
    snr_th = np.sqrt(mean_counts)

    measured_snr.append(snr_meas)
    theory_snr.append(snr_th)
    N_mean_list.append(mean_counts)

    # Save a few images for visualization
    noisy_examples.append((N0, img_noisy))

# ----------------------------
# 4) Plot images at different dose levels
# ----------------------------
fig, axes = plt.subplots(2, 4, figsize=(16, 8))
axes = axes.ravel()

axes[0].imshow(phantom, cmap='gray', vmin=0, vmax=1)
axes[0].set_title("Phantom (noise-free)")
axes[0].axis("off")

# show first 7 noisy examples (or fewer)
for i in range(1, min(8, len(noisy_examples) + 1)):
    N0, img_noisy = noisy_examples[i-1]
    axes[i].imshow(img_noisy, cmap='gray', vmin=0, vmax=1)
    axes[i].set_title(f"Poisson noise\nN0={N0}")
    axes[i].axis("off")

plt.tight_layout()
plt.show()

# ----------------------------
# 5) Verify SNR ~ sqrt(N): plot measured vs theory
# ----------------------------
plt.figure(figsize=(7,5))
plt.plot(N_mean_list, measured_snr, marker='o', label="Measured SNR (mean/std in ROI)")
plt.plot(N_mean_list, theory_snr, marker='s', label="Theory: sqrt(N)")
plt.xscale("log")
plt.yscale("log")
plt.xlabel("Mean photon count N in ROI (log scale)")
plt.ylabel("SNR (log scale)")
plt.title("Verification of SNR ≈ sqrt(N) for Poisson noise")
plt.grid(True, which="both")
plt.legend()
plt.show()

# Print a small table
print("N0\tMean N (ROI)\tMeasured SNR\tTheory sqrt(N)")
for N0, Nmean, s_meas, s_th in zip(N0_list, N_mean_list, measured_snr, theory_snr):
    print(f"{N0}\t{Nmean:.1f}\t\t{s_meas:.2f}\t\t{s_th:.2f}")
```

### **Coding Exercise 4**

Create a 2D analysis showing how image quality metrics vary with both radiation dose and detector pixel size.

You will create a heatmap showing the "detectability index" (combining SNR and resolution) as a function of:
- X-axis: Radiation dose (relative to baseline)
- Y-axis: Detector pixel size

This helps visualize the complex trade-offs in imaging system design.

```python
import numpy as np
import matplotlib.pyplot as plt

# Define parameter ranges
dose_range = np.logspace(-1, 1, 20)  # 0.1x to 10x baseline dose
pixel_size_range = np.linspace(0.1, 2.0, 20)  # 0.1 to 2.0 mm pixels

# Target object size (mm)
target_size = 0.5  # mm

# Initialize detectability array
detectability = np.zeros((len(pixel_size_range), len(dose_range)))

for i, pixel_size in enumerate(pixel_size_range):
    for j, dose in enumerate(dose_range):
        # Calculate detectability
        # SNR component from dose (assuming SNR is proportional to sqrt(dose))
        snr_component = np.sqrt(dose)

        # Resolution component (decreases as pixel size increases relative to target_size)
        # Simple model: if pixel_size is much smaller than target_size/2, resolution is maximal (1.0)
        # otherwise, it degrades inversely with pixel_size relative to target_size
        if pixel_size < target_size / 2:
            resolution_component = 1.0
        else:
            # As pixel_size increases, resolution_component should decrease.
            # A simple inverse relationship capped by the ideal resolution
            resolution_component = target_size / (2 * pixel_size)  # Decreases from 1.0 when pixel_size = target_size/2
            # Ensure resolution_component doesn't go below 0 (though unlikely with positive pixel_size)
            resolution_component = max(0.0, resolution_component)

        detectability[i, j] = snr_component * resolution_component

# Create heatmap
plt.figure(figsize=(12, 8))
plt.imshow(detectability, aspect='auto', origin='lower', cmap='viridis',
           extent=[dose_range[0], dose_range[-1], pixel_size_range[0], pixel_size_range[-1]])
plt.colorbar(label='Detectability Index')
plt.xlabel('Relative Dose', fontsize=12)
plt.ylabel('Pixel Size (mm)', fontsize=12)
plt.title(f'Image Quality Trade-offs\n(Target object size: {target_size} mm)', fontsize=14)
plt.xscale('log')

# Add contour lines
contours = plt.contour(dose_range, pixel_size_range, detectability,
                       levels=5, colors='white', alpha=0.5, linewidths=1)
plt.clabel(contours, inline=True, fontsize=8)

# Mark some interesting operating points
plt.plot([1.0], [0.5], 'r*', markersize=15, label='Baseline')
plt.legend(fontsize=10)

plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ----------------------------
# Parameters
# ----------------------------
dose_range = np.logspace(-1, 1, 30)          # 0.1x to 10x baseline
pixel_size_range = np.linspace(0.1, 2.0, 30) # mm
target_size = 0.5                             # mm (object diameter or characteristic size)

# Detectability index matrix
detectability = np.zeros((len(pixel_size_range), len(dose_range)))

# ----------------------------
# Model choices (simple but physics-consistent)
# - SNR increases with sqrt(dose)
# - Resolution penalty increases as pixel size approaches/exceeds target scale
#   Use a smooth sampling-limited penalty based on a Gaussian MTF approximation:
#   MTF(f) ~ exp(-(pi*sigma*f)^2), with sigma ~ pixel_size/sqrt(12) (uniform pixel aperture blur)
#   Choose f_target ~ 1/target_size (cycles/mm) as a characteristic spatial frequency.
# ----------------------------
for i, p in enumerate(pixel_size_range):
    sigma_aperture = p / np.sqrt(12)          # mm (std dev of box/pixel aperture blur)
    f_target = 1.0 / target_size              # cycles/mm (simple characteristic freq)

    # Resolution component ~ MTF at the target frequency (bounded 0..1, smooth)
    resolution_component = np.exp(-(np.pi * sigma_aperture * f_target)**2)

    for j, dose in enumerate(dose_range):
        snr_component = np.sqrt(dose)         # quantum-limited SNR scaling
        detectability[i, j] = snr_component * resolution_component

# ----------------------------
# Plot heatmap
# ----------------------------
plt.figure(figsize=(12, 8))

# extent sets the axis scales to physical units
plt.imshow(
    detectability,
    aspect='auto',
    origin='lower',
    extent=[dose_range[0], dose_range[-1], pixel_size_range[0], pixel_size_range[-1]]
)
plt.colorbar(label='Detectability Index (SNR × Resolution)')

plt.xscale('log')
plt.xlabel('Relative Dose (log scale)')
plt.ylabel('Pixel Size (mm)')
plt.title(f'Image Quality Trade-offs Heatmap\n(Target object size = {target_size} mm)')

# Contours
contours = plt.contour(dose_range, pixel_size_range, detectability, levels=6, linewidths=1)
plt.clabel(contours, inline=True, fontsize=8)

# Baseline operating point
plt.plot([1.0], [0.5], 'r*', markersize=15, label='Baseline (Dose=1, Pixel=0.5 mm)')
plt.legend()

plt.grid(True, alpha=0.25)
plt.tight_layout()
plt.show()
```

A basic sampling requirement in medical imaging is that an object should be sampled by at least 2 pixels across its diameter, with 3–6 pixels preferred to preserve shape and contrast and reduce partial-volume effects.

Pixels across a 3 mm nodule:

0.5 mm pixels → 3 / 0.5 = 6 pixels

1.0 mm pixels → 3 / 1.0 = 3 pixels

1.5 mm pixels → 3 / 1.5 = 2 pixels

Choice: 0.5 mm pixel size

Justification:
A 0.5 mm detector provides approximately 6 samples across a 3 mm nodule, which preserves edge information, minimizes partial-volume blurring, and provides a safety margin for detectability.
The 1.0 mm option is minimally acceptable, and 1.5 mm is borderline for reliable detection.

(b) Minimum photon count required to achieve SNR = 6

For quantum-limited (Poisson) noise:

Noise standard deviation = square root of N

Signal-to-noise ratio (SNR) = N / sqrt(N) = sqrt(N)

Set required SNR = 6:

6 = sqrt(N)

N = 36

Minimum required detected primary photon count (per measurement region):

N = 36 photons

(c) Effect of 40% Compton scatter on SNR and grid decision

If 40% of detected photons are Compton-scattered:

Primary (useful) photons = 60% of total

Scatter contributes noise but little useful contrast

Let total detected photons = N_total

Primary photons = 0.6 × N_total

Noise is driven by total counts → noise ≈ sqrt(N_total)

Effective SNR becomes:

SNR ≈ (0.6 × N_total) / sqrt(N_total)
SNR ≈ 0.6 × sqrt(N_total)

Set required SNR = 6:

0.6 × sqrt(N_total) = 6

sqrt(N_total) = 10

N_total = 100

So with 40% scatter:

About 100 total detected photons are required to achieve SNR = 6

Primary photons in this case ≈ 60.

Yes, an anti-scatter grid would likely be used.
Reason: Lung nodule detection is a low-contrast task, and a 40% scatter fraction significantly reduces effective contrast and SNR. Although a grid reduces primary photons and may require increased dose, it often provides a net improvement in detectability for this clinical task.
